 # Part 1 — Customers

Q1. Total customers kitne hain?

Q2. City-wise customers ki count nikalo.

Q3. Gender-wise customers ki count nikalo.

Q4. Customers ki average age kya hai?

Q5. Kaunsi city me sabse zyada customers hain?

```python
import pandas as pd
```

```python
ct=pd.read_csv("/content/customers.csv")
ct.head(5)
```

```python
od=pd.read_csv("/content/orders.csv")
od.head(5)
```

```python
pt=pd.read_csv("/content/products.csv")
pt.head(5)
```

```python
total_customers=ct['customer_id'].nunique()
print("total_customer", total_customers)
```

```python
city_customers=ct['city'].value_counts()
print("city_customers",city_customers)
```

```python
Gender_customers=ct['gender'].value_counts()
print("Gender_customers",Gender_customers)
```

```python
Average_age=ct['age'].mean()
print("Average_age",round(Average_age,2))
```

```python
maximum_customers=ct['city'].max()
print("maximum_customers",maximum_customers)
```

# Part 2 — Products

Q6. Total products kitne hain aur category-wise products kitne hain?

Q7. Har category ka average product price nikalo.

Q8. Top 5 most expensive products nikalo.

Q9. Kis category ke products ka average price sabse zyada hai?

```python
total_products=pt['product_id'].nunique()
print("total_products", total_products)
```

```python
category_wise_product=pt['category'].value_counts()
print("category_wise_product",category_wise_product)
```

```python
pt.groupby('category')['cost_price'].mean()
```

```python
most_expensive_products=pt.sort_values('cost_price',ascending=False).head(5)
most_expensive_products
```

```python
Avg_price=pt.groupby('category')['unit_price'].mean()
top_category=Avg_price.idxmax()
print("Highest Average Price Category: ",top_category)
```

# Part 3 — Orders

Q11. Delivered, Cancelled, Returned aur Pending orders ki count nikalo.

```python
od.columns
```

```python
Total_orders=od['order_id'].nunique()
print("Total_orders",Total_orders)
```

```python
order_status=od['order_status'].value_counts()
print("order_status: ",order_status)
```

```python
total_amount=od['total_amount'].value_counts()
print('total_amount',total_amount)
```
