```python

from datetime import datetime

parking_slots = 10
parked_vehicles = {}

while True:
    print("===== VEHICLE PARKING MANAGEMENT =====")
    print("1. Park Vehicle")
    print("2. View Parked Vehicles")
    print("3. Search Vehicle")
    print("4. Remove Vehicle")
    print("5. Check Available Slots")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        if len(parked_vehicles) < parking_slots:
            number = input("Enter vehicle number: ").upper()

            if number in parked_vehicles:
                print("Vehicle is already parked!")

            else:
                vehicle_type = input("Enter vehicle type (Car/Bike): ")

                entry_time = datetime.now()

                parked_vehicles[number] = {
                    "type": vehicle_type,
                    "entry": entry_time
                }

                print("Vehicle parked successfully!")
                print("Entry Time:", entry_time.strftime("%H:%M:%S"))

        else:
            print("Parking is full!")

    elif choice == "2":
        print("----- PARKED VEHICLES -----")

        if not parked_vehicles:
            print("No vehicles parked.")

        else:
            for number, details in parked_vehicles.items():
                print("Vehicle Number:", number)
                print("Vehicle Type:", details["type"])
                print("Entry Time:", details["entry"].strftime("%H:%M:%S"))
                print("---------------------")

    elif choice == "3":
        number = input("Enter vehicle number: ").upper()

        if number in parked_vehicles:
            print("Vehicle Found!")
            print("Vehicle Type:", parked_vehicles[number]["type"])
            print("Entry Time:", parked_vehicles[number]["entry"])
        else:
            print("Vehicle not found!")

    elif choice == "4":
        number = input("Enter vehicle number: ").upper()

        if number in parked_vehicles:
            entry_time = parked_vehicles[number]["entry"]
            exit_time = datetime.now()

            duration = exit_time - entry_time
            hours = max(1, (duration.total_seconds() + 3599) // 3600)

            vehicle_type = parked_vehicles[number]["type"].lower()

            if vehicle_type == "bike":
                rate = 20
            else:
                rate = 50

            bill = int(hours * rate)

            print("----- PARKING BILL -----")
            print("Vehicle:", number)
            print("Parking Hours:", int(hours))
            print("Rate per Hour: Rs.", rate)
            print("Total Bill: Rs.", bill)

            del parked_vehicles[number]

            print("Vehicle removed successfully!")

        else:
            print("Vehicle not found!")

    elif choice == "5":
        available = parking_slots - len(parked_vehicles)

        print("Total Parking Slots:", parking_slots)
        print("Occupied Slots:", len(parked_vehicles))
        print("Available Slots:", available)

    elif choice == "6":
        print("Thank you for using Parking Management System!")
        break

    else:
        print("Invalid choice! Please try again.")
```
