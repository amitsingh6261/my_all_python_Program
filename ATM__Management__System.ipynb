```python

accounts = {
    "1001": {
        "name": "Amit",
        "pin": "1234",
        "balance": 10000,
        "history": []
    },
    "1002": {
        "name": "Rahul",
        "pin": "5678",
        "balance": 15000,
        "history": []
    }
}

while True:
    print("\n===== ATM MANAGEMENT SYSTEM =====")

    account_no = input("Enter Account Number: ")
    pin = input("Enter ATM PIN: ")

    if account_no in accounts and accounts[account_no]["pin"] == pin:

        print("\nWelcome", accounts[account_no]["name"])

        while True:
            print("\n1. Check Balance")
            print("2. Deposit Money")
            print("3. Withdraw Money")
            print("4. Transaction History")
            print("5. Change PIN")
            print("6. Logout")

            choice = input("Enter your choice: ")

            if choice == "1":
                print("Available Balance: Rs.",
                      accounts[account_no]["balance"])

            elif choice == "2":
                amount = float(input("Enter deposit amount: "))

                if amount > 0:
                    accounts[account_no]["balance"] += amount
                    accounts[account_no]["history"].append(
                        "Deposited Rs. " + str(amount)
                    )
                    print("Money deposited successfully!")

                else:
                    print("Enter a valid amount!")

            elif choice == "3":
                amount = float(input("Enter withdrawal amount: "))

                if amount <= 0:
                    print("Enter a valid amount!")

                elif amount <= accounts[account_no]["balance"]:
                    accounts[account_no]["balance"] -= amount
                    accounts[account_no]["history"].append(
                        "Withdrawn Rs. " + str(amount)
                    )
                    print("Please collect your cash.")

                else:
                    print("Insufficient balance!")

            elif choice == "4":
                print("\n----- TRANSACTION HISTORY -----")

                history = accounts[account_no]["history"]

                if not history:
                    print("No transactions found.")

                else:
                    for transaction in history:
                        print(transaction)

            elif choice == "5":
                old_pin = input("Enter old PIN: ")

                if old_pin == accounts[account_no]["pin"]:
                    new_pin = input("Enter new 4-digit PIN: ")

                    if new_pin.isdigit() and len(new_pin) == 4:
                        accounts[account_no]["pin"] = new_pin
                        print("PIN changed successfully!")
                    else:
                        print("PIN must contain exactly 4 digits.")

                else:
                    print("Incorrect old PIN!")

            elif choice == "6":
                print("Successfully logged out!")
                break

            else:
                print("Invalid choice!")

    else:
        print("Invalid Account Number or PIN!")

    again = input("\nDo you want to use another account? (yes/no): ")

    if again.lower() != "yes":
        print("Thank you for using ATM!")
        break
```
