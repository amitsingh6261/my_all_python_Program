```python
class Car:

    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def show_details(self):
        print("--- Car Details ---")
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Price:", self.price)

    def start(self):
        print(self.brand, self.model, "is started.")

    def stop(self):
        print(self.brand, self.model, "is stopped.")

    def discount(self, amount):
        self.price = self.price - amount
        print("Discount:", amount)
        print("New Price:", self.price)


# Creating object
car1 = Car("Toyota", "Fortuner", 4000000)

# Calling methods
car1.show_details()

car1.start()

car1.discount(500000)

car1.stop()

# Final details
car1.show_details()
```
