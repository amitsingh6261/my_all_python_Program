# Filtering

# data filtering is the prosecc of selecting only the required data from a dataser based on the given condtion

```python
dt=pd.DataFrame({
     "Name":['Amit','Rohit','Nitin','Ram','Shyam','Naman'],
     "Age":[21,43,54,23,54,65],
     "City":['Indore','pune','delhi','agra','noida','devash'],
     "Marks":[99,78,98,96,85,93]
 })
```

```python
dt
```

```python
dt['Marks']>95
```

```python
dt[dt['Marks']>95]
```

3. Filtering ka Basic Formula

Isko yaad kar lo:

df[df["column"] condition]

AND → &

OR  → |



```python
dt[dt['Age']>45]
```

```python
#99 ko chhod kar
dt[dt['Marks']!=99]
```

```python
dt[dt['City']=='pune']
```

```python
dt[(dt['Age']>45)&(dt['Marks']>90)]
```

```python
dt[(dt['City']=='pune')|(dt['Marks']>90)]
```

```python
df[df["Marks"] > 80]["Name"]
```

DELETE

Delete ke do main cases hain:

Column delete karna hai.

Sorting

Sorting ka matlab:

Data ko kisi column ke according order me arrange karna.,,,,,

sort_values() ka basic syntax\\\\\\\\
df.sort_values("column_name")

\\\\Default me ascending order hota hai.

####
Important point:

df.sort_values("Marks")

sirf sorted result return karta hai. Original df ko permanently change nahi karta.

Agar original DataFrame ko hi sort karna hai:

df.sort_values("Marks", inplace=True)

Ab df permanently sorted ho jayega.

#############
df = df.sort_values("Marks").reset_index(drop=True)
drop=True ka matlab:

Purane index ko new column mat banao.
###
ascending=True   → छोटा → बड़ा
ascending=False  → बड़ा → छोटा


```python
df.sort_values('Score')
```

```python
df.sort_values('Age',ascending=False,inplace=True)
df.reset_index(drop=True,inplace=True)
df
```

```python
df.sort_values([
    'Score',
    'Student name'],ascending=True,inplace=True)
df.reset_index(drop=True,inplace=True)
df
```
