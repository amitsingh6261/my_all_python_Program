```python

import json
import os

FILE_NAME = "blood_bank.json"


def load_data():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)

    return {"donors": [], "blood_stock": {}}


def save_data():
    with open(FILE_NAME, "w") as file:
        json.dump(data, file, indent=4)


def add_donor():
    name = input("Enter donor name: ")
    age = int(input("Enter donor age: "))
    phone = input("Enter phone number: ")
    blood_group = input("Enter blood group: ").upper()

    donor = {
        "name": name,
        "age": age,
        "phone": phone,
        "blood_group": blood_group
    }

    data["donors"].append(donor)
    save_data()

    print("Donor added successfully!")


def view_donors():
    print("----- DONOR RECORDS -----")

    if not data["donors"]:
        print("No donor records found.")
        return

    for donor in data["donors"]:
        print("Name:", donor["name"])
        print("Age:", donor["age"])
        print("Phone:", donor["phone"])
        print("Blood Group:", donor["blood_group"])
        print("----------------------")


def search_blood():
    group = input("Enter blood group: ").upper()

    print("----- SEARCH RESULTS -----")

    found = False

    for donor in data["donors"]:
        if donor["blood_group"] == group:
            print("Donor:", donor["name"])
            print("Phone:", donor["phone"])
            found = True

    if not found:
        print("No donor found for this blood group.")


def add_blood_stock():
    group = input("Enter blood group: ").upper()
    units = int(input("Enter donated units: "))

    if units <= 0:
        print("Units must be greater than zero.")
        return

    data["blood_stock"][group] = (
        data["blood_stock"].get(group, 0) + units
    )

    save_data()
    print("Blood stock updated successfully!")


def view_blood_stock():
    print("----- AVAILABLE BLOOD STOCK -----")

    if not data["blood_stock"]:
        print("No blood stock available.")
        return

    for group, units in data["blood_stock"].items():
        print(group, ":", units, "units")


data = load_data()

while True:
    print("===== BLOOD BANK MANAGEMENT SYSTEM =====")
    print("1. Add Donor")
    print("2. View Donors")
    print("3. Search Blood Group")
    print("4. Add Blood Stock")
    print("5. View Blood Stock")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_donor()

    elif choice == "2":
        view_donors()

    elif choice == "3":
        search_blood()

    elif choice == "4":
        add_blood_stock()

    elif choice == "5":
        view_blood_stock()

    elif choice == "6":
        print("Thank you for using Blood Bank System!")
        break

    else:
        print("Invalid choice! Try again.")
```
