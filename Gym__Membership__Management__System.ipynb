```python

import json
import os
from datetime import date, timedelta

FILE_NAME = "gym_members.json"

plans = {
    1: {"name": "Monthly", "price": 1000, "days": 30},
    2: {"name": "Quarterly", "price": 2700, "days": 90},
    3: {"name": "Yearly", "price": 9000, "days": 365}
}


def load_members():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    return {}


def save_members():
    with open(FILE_NAME, "w") as file:
        json.dump(members, file, indent=4)


def register_member():
    member_id = input("Enter Member ID: ")

    if member_id in members:
        print("Member ID already exists!")
        return

    name = input("Enter member name: ")
    age = int(input("Enter age: "))
    phone = input("Enter phone number: ")

    print("Select Membership Plan:")
    for key, plan in plans.items():
        print(key, plan["name"], "- Rs.", plan["price"])

    choice = int(input("Enter plan number: "))

    if choice not in plans:
        print("Invalid plan!")
        return

    plan = plans[choice]
    start_date = date.today()
    end_date = start_date + timedelta(days=plan["days"])

    members[member_id] = {
        "name": name,
        "age": age,
        "phone": phone,
        "plan": plan["name"],
        "fee": plan["price"],
        "start_date": str(start_date),
        "end_date": str(end_date)
    }

    save_members()
    print("Member registered successfully!")
    print("Membership valid until:", end_date)


def view_members():
    print("----- GYM MEMBERS -----")

    if not members:
        print("No members found.")
        return

    for member_id, member in members.items():
        print("Member ID:", member_id)
        print("Name:", member["name"])
        print("Age:", member["age"])
        print("Plan:", member["plan"])
        print("Fee: Rs.", member["fee"])
        print("Expiry Date:", member["end_date"])


def search_member():
    member_id = input("Enter Member ID: ")

    if member_id in members:
        member = members[member_id]

        print("Member Found!")
        print("Name:", member["name"])
        print("Phone:", member["phone"])
        print("Plan:", member["plan"])
        print("Expiry:", member["end_date"])
    else:
        print("Member not found!")


def renew_membership():
    member_id = input("Enter Member ID: ")

    if member_id not in members:
        print("Member not found!")
        return

    print("Select New Plan:")
    for key, plan in plans.items():
        print(key, plan["name"], "- Rs.", plan["price"])

    choice = int(input("Enter plan number: "))

    if choice not in plans:
        print("Invalid plan!")
        return

    plan = plans[choice]
    old_end_date = date.fromisoformat(members[member_id]["end_date"])
    start_date = max(date.today(), old_end_date)
    end_date = start_date + timedelta(days=plan["days"])

    members[member_id]["plan"] = plan["name"]
    members[member_id]["fee"] = plan["price"]
    members[member_id]["end_date"] = str(end_date)

    save_members()

    print("Membership renewed successfully!")
    print("New expiry date:", end_date)


def delete_member():
    member_id = input("Enter Member ID: ")

    if member_id in members:
        del members[member_id]
        save_members()
        print("Member deleted successfully!")
    else:
        print("Member not found!")


members = load_members()

while True:
    print("===== GYM MANAGEMENT SYSTEM =====")
    print("1. Register Member")
    print("2. View Members")
    print("3. Search Member")
    print("4. Renew Membership")
    print("5. Delete Member")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        register_member()

    elif choice == "2":
        view_members()

    elif choice == "3":
        search_member()

    elif choice == "4":
        renew_membership()

    elif choice == "5":
        delete_member()

    elif choice == "6":
        print("Thank you for using Gym Management System!")
        break

    else:
        print("Invalid choice! Try again.")
```
