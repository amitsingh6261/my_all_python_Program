```python

questions = [
    {
        "question": "1. Which language is used for AI?",
        "options": ["A. HTML", "B. Python", "C. CSS", "D. XML"],
        "answer": "B"
    },
    {
        "question": "2. What is 10 + 20?",
        "options": ["A. 20", "B. 25", "C. 30", "D. 40"],
        "answer": "C"
    },
    {
        "question": "3. Which data type stores True or False?",
        "options": ["A. String", "B. List", "C. Float", "D. Boolean"],
        "answer": "D"
    },
    {
        "question": "4. Which symbol is used for comments in Python?",
        "options": ["A. #", "B. //", "C. @", "D. $"],
        "answer": "A"
    },
    {
        "question": "5. Which function displays output?",
        "options": ["A. input()", "B. print()", "C. type()", "D. int()"],
        "answer": "B"
    }
]


def start_quiz():
    score = 0

    print("\n===== PYTHON QUIZ COMPETITION =====")
    name = input("Enter your name: ")

    for question in questions:
        print("\n" + question["question"])

        for option in question["options"]:
            print(option)

        answer = input("Enter your answer (A/B/C/D): ").upper()

        if answer == question["answer"]:
            print("Correct answer!")
            score += 1
        else:
            print("Wrong answer!")

    percentage = (score / len(questions)) * 100

    print("\n===== QUIZ RESULT =====")
    print("Participant:", name)
    print("Total Questions:", len(questions))
    print("Correct Answers:", score)
    print("Wrong Answers:", len(questions) - score)
    print("Percentage:", percentage, "%")

    if percentage >= 80:
        print("Result: Excellent!")
    elif percentage >= 60:
        print("Result: Good!")
    elif percentage >= 40:
        print("Result: Keep Practicing!")
    else:
        print("Result: Try Again!")


while True:
    print("\n1. Start Quiz")
    print("2. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        start_quiz()
    elif choice == "2":
        print("Thank you for participating!")
        break
    else:
        print("Invalid choice!")
```
