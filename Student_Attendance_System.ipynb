```python

students = {}


def add_student():
    roll_no = input("Enter roll number: ")
    name = input("Enter student name: ")

    if roll_no in students:
        print("Student already exists!")
    else:
        students[roll_no] = {
            "name": name,
            "present": 0,
            "absent": 0
        }
        print("Student added successfully!")


def mark_attendance():
    roll_no = input("Enter roll number: ")

    if roll_no not in students:
        print("Student not found!")
        return

    status = input("Enter attendance (P/A): ").upper()

    if status == "P":
        students[roll_no]["present"] += 1
        print("Present marked successfully!")
    elif status == "A":
        students[roll_no]["absent"] += 1
        print("Absent marked successfully!")
    else:
        print("Invalid choice! Enter P or A.")


def view_students():
    if not students:
        print("No student records found!")
        return

    print("===== STUDENT ATTENDANCE =====")

    for roll_no, details in students.items():
        total = details["present"] + details["absent"]

        if total > 0:
            percentage = details["present"] / total * 100
        else:
            percentage = 0

        print("Roll Number:", roll_no)
        print("Name:", details["name"])
        print("Present:", details["present"])
        print("Absent:", details["absent"])
        print("Attendance:", round(percentage, 2), "%")
        print("---------------------------")


def search_student():
    roll_no = input("Enter roll number to search: ")

    if roll_no in students:
        print("Name:", students[roll_no]["name"])
        print("Present:", students[roll_no]["present"])
        print("Absent:", students[roll_no]["absent"])
    else:
        print("Student not found!")


while True:
    print("===== ATTENDANCE MANAGEMENT =====")
    print("1. Add Student")
    print("2. Mark Attendance")
    print("3. View All Students")
    print("4. Search Student")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()
    elif choice == "2":
        mark_attendance()
    elif choice == "3":
        view_students()
    elif choice == "4":
        search_student()
    elif choice == "5":
        print("Program closed!")
        break
    else:
        print("Invalid choice!")
```
