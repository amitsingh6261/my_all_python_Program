#  loc vs iloc

# loc = index/label dekhta hai (kis index per value hai)
# iloc = position dekhta hai(kis possion per value hai)
#loc → Label/Index aur column ke naam par kaam karta hai.
#Example: df.loc[1, "Name"]
#iloc → Integer position (0, 1, 2...) par kaam karta hai.
#Example: df.iloc[1, 0]

# Filtering

# data filtering is the prosecc of selecting only the required data from a dataser based on the given condtion

```python
dt=pd.DataFrame({
     "Name":['Amit','Rohit','Nitin','Ram','Shyam','Naman'],
     "Age":[21,43,54,23,54,65],
     "City":['Indore','pune','delhi','agra','noida','devash'],
     "Marks":[99,78,98,96,85,93]
 })
```

```python
dt
```

```python
dt['Marks']>95
```

```python
dt[dt['Marks']>95]
```

3. Filtering ka Basic Formula

Isko yaad kar lo:

df[df["column"] condition]

AND → &

OR  → |



```python
dt[dt['Age']>45]
```

```python
#99 ko chhod kar
dt[dt['Marks']!=99]
```

```python
dt[dt['City']=='pune']
```

```python
dt[(dt['Age']>45)&(dt['Marks']>90)]
```

```python
dt[(dt['City']=='pune')|(dt['Marks']>90)]
```

```python
df[df["Marks"] > 80]["Name"]
```
