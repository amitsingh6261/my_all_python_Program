# Level 2 - Filtering

.₹60,000 se zyada salary wale employees nikalo.

.IT department ke employees nikalo.

.5+ years experience wale employees nikalo.

.Performance score 80+ wale employees nikalo.

.Bhopal ke employees jinki salary ₹50,000+ hai, unhe nikalo.

.IT department ke employees jinki performance 85+ hai.

```python
df[df['Monthly_Salary']>60000]
```

```python
df[df['Department']=='IT']
```

```python
df[df['Experience_Years']>5]
```

```python
df[df['Performance_Score']>80]
```

```python
df[(df['City']=='Bhopal') & (df['Monthly_Salary']>50000)]
```

```python
df[(df['Department']=='IT') & (df['Performance_Score']>85)]
```

# Level 3 — Add & Modify
.Salary_Level naam ka column banao.

.Salary >= 60000 → "High"


.Salary < 60000 → "Normal"

.Performance_Level column banao.

.Kisi specific employee ki salary modify karo.

.Kisi condition ke according data modify karo.

```python
df['Salary_Level']='Normal'
```

```python
df
```

```python
df.loc[df['Monthly_Salary']>=60000, 'Salary_Level']="High"
df
```

```python
df['Performance_Level']='Average'
df.loc[df['Performance_Score']>75,'Performance_Level']='Good'
```

```python
df
```

```python
df.loc[df['Employee_ID']==1001,'Monthly_Salary']=55000
df
```

```python
df.loc[df['Age']<=28,'Age']+=5
df
```

# Level 4 — Delete & Rename
.Ek unnecessary column delete karo.

.Name ko Employee_Name rename karo.

.Monthly_Salary ko Salary rename karo.

```python
df.drop('Employee_ID',axis=1,inplace=True)
df
```

```python
df.rename(columns={
    'Name':"Employee_Name",
    'Monthly_Salary':'Salary'
},inplace=True)
df
```

# Level 5 — Sorting
.Salary ko highest → lowest sort karo.

.Experience ko highest → lowest sort karo.



```python
df.sort_values(
    ["Salary", "Experience_Years",],
    ascending=[False, False],
    inplace=True
)
df
```

```python
df.to_csv("/content/final_employee_report.csv",index=False)
df
```
