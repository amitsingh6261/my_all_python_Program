```python
# Student Result Management System

class Student:

    def __init__(self, student_id, name, course):
        self.student_id = student_id
        self.name = name
        self.course = course
        self.marks = {}

    def add_marks(self):
        print("\nEnter marks out of 100")

        self.marks["Python"] = float(input("Python: "))
        self.marks["Maths"] = float(input("Maths: "))
        self.marks["DBMS"] = float(input("DBMS: "))
        self.marks["Computer"] = float(input("Computer: "))
        self.marks["English"] = float(input("English: "))

        print("Marks added successfully!")

    def total(self):
        return sum(self.marks.values())

    def percentage(self):
        return self.total() / len(self.marks)

    def grade(self):

        percentage = self.percentage()

        if percentage >= 90:
            return "A+"
        elif percentage >= 80:
            return "A"
        elif percentage >= 70:
            return "B"
        elif percentage >= 60:
            return "C"
        elif percentage >= 50:
            return "D"
        elif percentage >= 33:
            return "E"
        else:
            return "F"

    def result(self):

        for mark in self.marks.values():

            if mark < 33:
                return "Fail"

        return "Pass"

    def display(self):

        print("\n-----------------------------")
        print("Student ID :", self.student_id)
        print("Name       :", self.name)
        print("Course     :", self.course)

        print("\nMarks:")

        for subject, mark in self.marks.items():
            print(subject, ":", mark)

        if len(self.marks) > 0:
            print("\nTotal      :", self.total())
            print("Percentage :", self.percentage(), "%")
            print("Grade      :", self.grade())
            print("Result     :", self.result())

        print("-----------------------------")


class StudentManagement:

    def __init__(self):
        self.students = []

    def add_student(self):

        print("\n===== ADD STUDENT =====")

        student_id = input("Enter Student ID: ")

        for student in self.students:

            if student.student_id == student_id:
                print("Student ID already exists!")
                return

        name = input("Enter Student Name: ")
        course = input("Enter Course: ")

        student = Student(
            student_id,
            name,
            course
        )

        self.students.append(student)

        print("Student added successfully!")

    def view_students(self):

        print("\n===== ALL STUDENTS =====")

        if len(self.students) == 0:
            print("No students found.")
            return

        for student in self.students:
            student.display()

    def search_student(self):

        print("\n===== SEARCH STUDENT =====")

        student_id = input("Enter Student ID: ")

        for student in self.students:

            if student.student_id == student_id:
                student.display()
                return

        print("Student not found!")

    def add_marks(self):

        print("\n===== ADD MARKS =====")

        student_id = input("Enter Student ID: ")

        for student in self.students:

            if student.student_id == student_id:
                student.add_marks()
                return

        print("Student not found!")

    def delete_student(self):

        print("\n===== DELETE STUDENT =====")

        student_id = input("Enter Student ID: ")

        for student in self.students:

            if student.student_id == student_id:

                self.students.remove(student)

                print("Student deleted successfully!")
                return

        print("Student not found!")

    def find_topper(self):

        print("\n===== CLASS TOPPER =====")

        if len(self.students) == 0:
            print("No students found.")
            return

        topper = None

        for student in self.students:

            if len(student.marks) == 0:
                continue

            if topper is None:
                topper = student

            elif student.percentage() > topper.percentage():
                topper = student

        if topper is None:
            print("No marks available.")
            return

        print("\nTopper:")
        print("Name       :", topper.name)
        print("Student ID :", topper.student_id)
        print("Percentage :", topper.percentage(), "%")
        print("Grade      :", topper.grade())

    def menu(self):

        while True:

            print("\n================================")
            print("   STUDENT RESULT MANAGEMENT")
            print("================================")

            print("1. Add Student")
            print("2. View Students")
            print("3. Search Student")
            print("4. Add Marks")
            print("5. Delete Student")
            print("6. Find Topper")
            print("7. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":
                self.add_student()

            elif choice == "2":
                self.view_students()

            elif choice == "3":
                self.search_student()

            elif choice == "4":
                self.add_marks()

            elif choice == "5":
                self.delete_student()

            elif choice == "6":
                self.find_topper()

            elif choice == "7":
                print("Thank you!")
                break

            else:
                print("Invalid choice!")


# Main Program

system = StudentManagement()

system.menu()
```
