```python
# 14. Find the maximum and minimum element in a list (without using built-in functions).

min=999
max=-9999
l2=[44,66,89,73,42,5,9,88]
for i in l2:
  if i>max:
    max=i
  if i<min:
   min=i
print(max)
print(min)
```

```python
# 3. Remove duplicates from a list.
new_list=[]
l3=[4,5,6,73,4,6,2,9,8,5,6]
for i in l3:
  if i not in new_list:
    new_list.append(i)
print(new_list)
```

```python
# Take a number and check whether it is a palindrome.

list=[1,2,3,4,5]
print(list)
copy_list=list.copy()
copy_list.reverse()
print("reverse list: ",copy_list)

if copy_list==list:
  print("palindrome")
else:
  print("not palindrome")
```
