```python
rooms = [101, 102, 103, 104, 105]
booked = []

while True:

    print("\n--- HOTEL MANAGEMENT ---")
    print("1. Show Rooms")
    print("2. Book Room")
    print("3. Cancel Room")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":

        print("\nAvailable Rooms:")

        for room in rooms:

            if room not in booked:
                print(room)

    elif choice == "2":

        room = int(input("Enter room number: "))

        if room in rooms:

            if room not in booked:

                name = input("Enter customer name: ")

                booked.append(room)

                print("Room booked successfully!")
                print("Customer:", name)
                print("Room:", room)

            else:
                print("Room already booked.")

        else:
            print("Room does not exist.")

    elif choice == "3":

        room = int(input("Enter room number: "))

        if room in booked:

            booked.remove(room)

            print("Room cancelled successfully!")

        else:
            print("Room is not booked.")

    elif choice == "4":

        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
