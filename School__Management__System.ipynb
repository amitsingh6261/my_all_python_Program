```python


FILE_NAME = "school_data.json"


def load_data():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)

    return {"students": {}, "teachers": {}}


def save_data():
    with open(FILE_NAME, "w") as file:
        json.dump(data, file, indent=4)


def add_student():
    roll = input("Enter Roll Number: ")

    if roll in data["students"]:
        print("Roll number already exists!")
        return

    name = input("Enter Student Name: ")
    age = int(input("Enter Age: "))
    student_class = input("Enter Class: ")

    data["students"][roll] = {
        "name": name,
        "age": age,
        "class": student_class,
        "marks": {}
    }

    save_data()
    print("Student added successfully!")


def view_students():
    print("\n----- STUDENT RECORDS -----")

    if not data["students"]:
        print("No student records found.")
        return

    for roll, student in data["students"].items():
        print("\nRoll Number:", roll)
        print("Name:", student["name"])
        print("Age:", student["age"])
        print("Class:", student["class"])


def add_marks():
    roll = input("Enter Roll Number: ")

    if roll not in data["students"]:
        print("Student not found!")
        return

    student = data["students"][roll]

    for subject in ["Python", "Maths", "English"]:
        marks = int(input("Enter " + subject + " marks: "))

        if marks < 0 or marks > 100:
            print("Marks must be between 0 and 100.")
            return

        student["marks"][subject] = marks

    save_data()
    print("Marks added successfully!")


def view_result():
    roll = input("Enter Roll Number: ")

    if roll not in data["students"]:
        print("Student not found!")
        return

    student = data["students"][roll]
    marks = student["marks"]

    if len(marks) != 3:
        print("Marks are not available for all subjects.")
        return

    total = sum(marks.values())
    percentage = total / 3

    print("\n===== STUDENT RESULT =====")
    print("Name:", student["name"])

    for subject, score in marks.items():
        print(subject, ":", score)

    print("Total Marks:", total, "/ 300")
    print("Percentage:", percentage, "%")

    if percentage >= 40 and all(score >= 33 for score in marks.values()):
        print("Result: PASS")
    else:
        print("Result: FAIL")


def add_teacher():
    teacher_id = input("Enter Teacher ID: ")

    if teacher_id in data["teachers"]:
        print("Teacher ID already exists!")
        return

    name = input("Enter Teacher Name: ")
    subject = input("Enter Subject: ")

    data["teachers"][teacher_id] = {
        "name": name,
        "subject": subject
    }

    save_data()
    print("Teacher added successfully!")


def view_teachers():
    print("\n----- TEACHER RECORDS -----")

    if not data["teachers"]:
        print("No teacher records found.")
        return

    for teacher_id, teacher in data["teachers"].items():
        print("\nTeacher ID:", teacher_id)
        print("Name:", teacher["name"])
        print("Subject:", teacher["subject"])


data = load_data()

while True:
    print("\n===== SCHOOL MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Add Student Marks")
    print("4. View Student Result")
    print("5. Add Teacher")
    print("6. View Teachers")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        view_students()

    elif choice == "3":
        add_marks()

    elif choice == "4":
        view_result()

    elif choice == "5":
        add_teacher()

    elif choice == "6":
        view_teachers()

    elif choice == "7":
        print("Thank you for using School Management System!")
        break

    else:
        print("Invalid choice! Try again.")
```
