```python
class Student():
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def get_avg(self):
    sum=0
    for val in self.marks:
      sum+=val
    print(f"hii {self.name} your avg score is {sum/3}")

s1=Student("amit",[98,97,99])
s1.get_avg()
print(s1.name)
print(s1.marks)
```
