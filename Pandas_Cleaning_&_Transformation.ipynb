# Definition
Data cleaning and Transformation in pandas is the process of removing error,missing or duplicate data and changing tha data into a proper and useful format for analylis.

```python
import pandas as pd
df=pd.read_csv('/content/pandas_data_cleaning_1000.csv')
```

```python
df
```

```python
# df.isnull() DataFrame ke har cell mein check karta hai ki value missing hai ya nahi. ans true false me deta hai

df.isnull()
```

```python
# df.isna() ka kaam exactly df.isnull() jaisa hi hai.

df.isna()
```

```python
# Har column mein kitni missing values hain, ye batata hai.
df.isnull().sum()
```

```python
# df[df.isnull().any(axis=1)]
#Jis row mein kam se kam ek missing value hai, woh poori row dikhata hai.

df[df.isnull().any(axis=1)]
```

```python
df[df['Salary'].isna()]
```

```python
# Salary ka median use karke jaha missing value hogi waha fill kar dega.
df['Salary']=df['Salary'].fillna(df['Salary'].median())
df
```

```python
df.isnull().sum()
```

```python
df['City']=df['City'].fillna('Unknown')
df
```

```python
df.isnull().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
df
```

```python
df.isnull().sum()
```

```python
df['Performance_Score']=df['Performance_Score'].fillna(df['Performance_Score'].median())
df
```

```python
df['Experience_Years']=df['Experience_Years'].fillna(df['Experience_Years'].median())
```

```python
df.isna().sum()
```

# df.dropna()

 missing values ko delete karte hain.

Jis row mein ek bhi missing value hai, poori row delete.

Sirf Salary ki missing values wali rows delete karega.

# df.dropna(axis=1)

Ab rows nahi, columns delete karenge.

# df.ffill()

ffill = Forward Fill

Missing value ko upar wali previous value se fill karo.

# df.bfill()

bfill = Backward Fill

Missing value ko neeche wali next value se fill karo.

```python
# Age column ko integer (int64) mein convert karta hai.
# int64 missing NaN ko handle nahi karta.

df['Age']=df['Age'].astype('int64')
```

```python
df['Age'].dtypes
```

```python
#coerce ka simple meaning:
#Jo value number mein convert nahi ho sakti, usko NaN bana do.
# ye text or string data ko numeric data me convert krta hai


df['City']=pd.to_numeric(df['City'],errors='coerce')
```

```python
df['City'].dtypes
```

```python
#Iska kaam hai date ko proper Pandas date/time format mein convert karna

df['Date']=pd.to_datetime(df['Date'],errors='coerce')
```
